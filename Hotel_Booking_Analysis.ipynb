```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('hotel_bookings.csv')

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)
df.head()
df.info()
df.isnull().sum()
```

```python
print("========== HOTEL BOOKING DATASET REPORT ==========")

print("\n1. DATASET DIMENSIONS")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n2. COLUMN NAMES")
print(df.columns.tolist())

print("\n3. DATA TYPES")
print(df.dtypes)

print("\n4. MISSING VALUES")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\n5. DUPLICATE RECORDS")
print(df.duplicated().sum())

print("\n6. STATISTICAL SUMMARY")
display(df.describe(include="all").T)

print("\n7. FIRST FIVE RECORDS")
display(df.head())
```

```python
df[df['children'].isnull()]
```

```python
# Remove duplicate records
df = df.drop_duplicates()

# Handle missing values
df['children'] = df['children'].fillna(0)
df['country'] = df['country'].fillna('Unknown')
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)

# Convert date column safely
df['reservation_status_date'] = pd.to_datetime(
    df['reservation_status_date'],
    format='mixed',
    errors='coerce'
)

# Check results
print("Data cleaning completed!")
print("Final dataset shape:", df.shape)
print("\nRemaining missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print("\nDuplicate records:", df.duplicated().sum())
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Count bookings by hotel type
hotel_counts = df['hotel'].value_counts()

print("Hotel Booking Distribution:")
print(hotel_counts)

# Visualize hotel bookings
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='hotel')

plt.title('Booking Distribution by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Number of Bookings')
plt.show()
```

```python
# Calculate cancellation statistics
cancellation_counts = df['is_canceled'].value_counts()

print("Booking Cancellation Details:")
print("Not Cancelled:", cancellation_counts.get(0, 0))
print("Cancelled:", cancellation_counts.get(1, 0))

# Calculate cancellation percentage
cancellation_rate = df['is_canceled'].mean() * 100

print("Cancellation Rate:", round(cancellation_rate, 2), "%")

# Visualize cancellation
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='is_canceled')

plt.title('Booking Cancellation Analysis')
plt.xlabel('Cancellation (0 = No, 1 = Yes)')
plt.ylabel('Number of Bookings')
plt.show()
```

```python
# Define month order
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

# Count bookings by year and month
monthly_bookings = (
    df.groupby(['arrival_date_year', 'arrival_date_month'])
    .size()
    .reset_index(name='total_bookings')
)

monthly_bookings['arrival_date_month'] = pd.Categorical(
    monthly_bookings['arrival_date_month'],
    categories=month_order,
    ordered=True
)

monthly_bookings = monthly_bookings.sort_values(
    ['arrival_date_year', 'arrival_date_month']
)

display(monthly_bookings)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_bookings,
    x='arrival_date_month',
    y='total_bookings',
    hue='arrival_date_year',
    marker='o'
)

plt.title('Monthly Hotel Booking Trends')
plt.xlabel('Month')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.legend(title='Year')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Average Daily Rate by hotel type
adr_by_hotel = df.groupby('hotel')['adr'].mean()

print("Average Daily Rate by Hotel:")
print(adr_by_hotel.round(2))

# Average price by month
monthly_adr = df.groupby('arrival_date_month', observed=False)['adr'].mean()

monthly_adr = monthly_adr.reindex(month_order)

print("\nMonthly Average Daily Rate:")
print(monthly_adr.round(2))
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=monthly_adr.index,
    y=monthly_adr.values
)

plt.title('Average Daily Room Rate by Month')
plt.xlabel('Month')
plt.ylabel('Average Daily Rate')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=monthly_adr.index,
    y=monthly_adr.values
)

plt.title('Average Daily Room Rate by Month')
plt.xlabel('Month')
plt.ylabel('Average Daily Rate')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='hotel',
    y='adr',
    errorbar=None
)

plt.title('Average Daily Rate by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Average Daily Rate')
plt.show()
```

```python
# Top 10 countries by bookings
top_countries = (
    df[df['country'] != 'Unknown']['country']
    .value_counts()
    .head(10)
)

print("Top 10 Customer Countries:")
print(top_countries)

# Visualisation
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title('Top 10 Countries by Hotel Bookings')
plt.xlabel('Number of Bookings')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
# Key Performance Indicators (KPIs)

total_bookings = len(df)
cancelled_bookings = int(df['is_canceled'].sum())
successful_bookings = total_bookings - cancelled_bookings
cancellation_rate = (cancelled_bookings / total_bookings) * 100
average_adr = df['adr'].mean()

print("HOTEL BOOKING ANALYSIS - KPI SUMMARY")
print("------------------------------------")
print("Total Bookings:", total_bookings)
print("Successful Bookings:", successful_bookings)
print("Cancelled Bookings:", cancelled_bookings)
print("Cancellation Rate:", round(cancellation_rate, 2), "%")
print("Average Daily Rate:", round(average_adr, 2))
```

```python
import matplotlib.pyplot as plt

# Calculate KPI values
total_bookings = len(df)
cancelled_bookings = int(df['is_canceled'].sum())
successful_bookings = total_bookings - cancelled_bookings
cancellation_rate = df['is_canceled'].mean() * 100
average_adr = df['adr'].mean()

# Create KPI Dashboard
kpi_labels = [
    'Total Bookings',
    'Successful Bookings',
    'Cancelled Bookings',
    'Cancellation Rate',
    'Average ADR'
]

kpi_values = [
    total_bookings,
    successful_bookings,
    cancelled_bookings,
    round(cancellation_rate, 2),
    round(average_adr, 2)
]

fig, axes = plt.subplots(1, 5, figsize=(18, 4))

for ax, label, value in zip(axes, kpi_labels, kpi_values):
    ax.text(0.5, 0.6, f'{value:,}', ha='center',
            va='center', fontsize=18, fontweight='bold')
    ax.text(0.5, 0.3, label, ha='center',
            va='center', fontsize=10)
    ax.set_xticks([])
    ax.set_yticks([])
    for spine in ax.spines.values():
        spine.set_visible(True)

plt.suptitle('Hotel Booking Analysis - KPI Dashboard',
             fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# FINAL BUSINESS INSIGHTS

total_bookings = len(df)
cancellation_rate = df['is_canceled'].mean() * 100

top_hotel = df['hotel'].value_counts().idxmax()
top_country = df[df['country'] != 'Unknown']['country'].value_counts().idxmax()

peak_month = df['arrival_date_month'].value_counts().idxmax()

average_adr = df['adr'].mean()

print("HOTEL BOOKING ANALYSIS - BUSINESS INSIGHTS")
print("------------------------------------------")

print(f"1. The hotel type with the most bookings is: {top_hotel}")

print(f"2. The overall booking cancellation rate is: {cancellation_rate:.2f}%")

print(f"3. The country with the highest number of bookings is: {top_country}")

print(f"4. The most popular arrival month is: {peak_month}")

print(f"5. The average daily room rate (ADR) is: {average_adr:.2f}")

print("\nAnalysis completed successfully!")
```
